#### 조건 조회

- SQL WHERE 절과 동일

````python
df['score'] >= 90
````

````python
# 점수가 90점 이상
df[df['score'] >= 90]
````

- NumPy와 동일한 조건 연산
- 파이썬 : and, or, not 사용
- NumPy, Pandas : `&`, `|`, `~` 사용
- 연산자 우선순위로 각 조건은 반드시 `()`를 사용할 것

````python
# 점수가 90점 초과이고 나이가 21살 이상인 학생 데이터 참/거짓 여부
(df['score'] > 90) & (df['age'] >= 21) 
````

````python
df[(df['score'] > 90) & (df['age'] >= 21)]
````

````python
# 점수가 90 초과 이거나 나이가 21살 이상이거나 둘 중에 하나로 참인 데이터 조회
df[(df['score'] > 90) | (df['age'] >= 21)]
````

#### 컬럼 추가
- 사용하던 DF 이외에 필요한 컬럼이 생길때 추가

````python
df
````

````python
# 80점을 초과하는 학생만 통과
# 같은 이름 컬럼은 새로 추가 안됨
df['pass'] = df['score'] > 80

df
````

````python
df['score_bonus'] = df['score'] + 5

df
````

#### 컬럼 수정

- 기존 컬럼 수정

````python
df['score_bonus'] = df['score_bonus'] - 2

df
````

````python
# 행선택 에서 score_bonus 점수가 100점이 넘는 행만 선택 후, score_bonus 열의 값을 100으로 수정
df.loc[df['score_bonus'] > 100, 'score_bonus'] = 100

df
````

#### 컬럼 삭제

- 불필요한 컬럼은 삭제 

````python
df.drop(columns=['score_bonus'])
````

````python
# 두 번 실행하면 에러. 컬럼이 없어졌기 때문에
df.drop(columns=['score_bonus'], inplace=True)
````

### 실습

- 상품데이터 활용

````python
data = {
    'product': ['키보드', '마우스', '모니터', 'USB', '노트북'],
    'category': ['주변기기', '주변기기', '디스플레이', '저장장치', '컴퓨터'],
    'price': [30000, 15000, 250000, 12000, 1200000],
    'stock': [10, 30, 5, 50, 3]
}

df_products = pd.DataFrame(data)
````

1. 전체 데이터 출력
2. product, price 컬럼 출력
3. 가격이 50000원 이상인 상품만 출력
4. 재고가 10개 이상인 상품만 출력
5. total_value 컬럼 추가. 계산식은 price * stock
6. 전체 재고 가치의 합계 출력
7. 카테고리가 주변기기인 상품만 출력

````python
# 1번
df_products
````

````python
# 2번
df_products[['product', 'price']]
````

````python
# 3번
df_products['price'] >= 50000
````

````python
df_products[df_products['price'] >= 50000]
````

````python
# 4번
df_products[ df_products['stock'] >= 10 ]
````

````python
# 5번
df_products['total_value'] = df_products['price'] * df_products['stock']

df_products
````

````python
# 6번 
df_products['total_value'].sum()
````

````python
# 7번
df_products[ df_products['category'] == '주변기기' ]
````
